```python
LAST_NAME = "Lacdan"
STUDENT_ID = "TUPM-26-8415"
FAVORITE_ARTIST = "Sleeping At Last"
SEED_DIGIT = "5"
ID_SUM = "26"
NAME_LENGTH = "6"
ARTIST_LENGTH = "16"
ARTIST_SUM = "161"

# 1. Generate original and rotated records
n = 10 + int(NAME_LENGTH)
original = sorted(set(
    int(ID_SUM) * i + int(ARTIST_SUM) + int(SEED_DIGIT)
    for i in range(1, n + 1)
))
rotation = int(SEED_DIGIT) % len(original)
rotated = original[rotation:] + original[:rotation]

# 2. Generate target incident
target = int(ID_SUM) * int(ARTIST_LENGTH) + int(ARTIST_SUM)

# 3. Binary search in rotated list
left, right = 0, len(rotated) - 1
comparisons = 0
position = -1

while left <= right:
    mid = (left + right) // 2
    comparisons += 1

    if rotated[mid] == target:
        position = mid
        break

    if rotated[left] <= rotated[mid]:
        if rotated[left] <= target < rotated[mid]:
            right = mid - 1
        else:
            left = mid + 1
    else:
        if rotated[mid] < target <= rotated[right]:
            left = mid + 1
        else:
            right = mid - 1

# 4. Linear search comparison
linear_comparisons = 0
linear_position = -1

for i, record in enumerate(rotated):
    linear_comparisons += 1
    if record == target:
        linear_position = i
        break

# 5. Assessment output
print("Student:", LAST_NAME, "|", STUDENT_ID)
print("Original Data Set:", original)
print("Rotated Data Set:", rotated)
print("Target Incident:", target)
print("Binary Search Position:", position + 1 if position != -1 else "Not Found")
print("Binary Search Comparisons:", comparisons)
print("Linear Search Position:", linear_position + 1 if linear_position != -1 else "Not Found")
print("Linear Search Comparisons:", linear_comparisons)
print("Results Match:", position == linear_position)
print("Time Complexity: O(log n) average/worst case for this binary search")
print("Space Complexity: O(n) for storing the datasets; O(1) auxiliary search space")
print("Final Output:", "Incident Found" if position != -1 else "Incident Not Found")
```

```python

LAST_NAME = "Lacdan"
STUDENT_ID = "TUPM-26-8415"
FAVORITE_ARTIST = "Sleeping At Last"
SEED_DIGIT = "5"
ID_SUM = "26"
NAME_LENGTH = "6"
ARTIST_LENGTH = "16"
ARTIST_SUM = "161"

# 1. Generate sorted department datasets
n = 10
A = sorted(int(SEED_DIGIT) * i + int(ID_SUM) for i in range(1, n + 1))
B = sorted(int(ARTIST_LENGTH) * i + int(NAME_LENGTH) for i in range(1, n + 1))

# 2. Merge the datasets
merged = []
i = j = comparisons = 0

while i < len(A) and j < len(B):
    comparisons += 1
    if A[i] <= B[j]:
        merged.append(A[i])
        i += 1
    else:
        merged.append(B[j])
        j += 1

merged.extend(A[i:])
merged.extend(B[j:])

# 3. Verify results
ordered = all(merged[k] <= merged[k + 1] for k in range(len(merged) - 1))
record_count_correct = len(merged) == len(A) + len(B)

print("Student:", LAST_NAME, "|", STUDENT_ID)
print("Department A Dataset:", A)
print("Department B Dataset:", B)
print("Consolidated Dataset:", merged)
print("Records Before:", len(A) + len(B))
print("Records After:", len(merged))
print("Comparisons:", comparisons)
print("Ordering Verified:", ordered)
print("Record Count Verified:", record_count_correct)
print("Time Complexity: O(n + m)")
print("Space Complexity: O(n + m)")
print("Final Output:", "Consolidation Successful" if ordered and record_count_correct else "Verification Failed")
```

```python
import random
from time import perf_counter

LAST_NAME = "Lacdan"
STUDENT_ID = "TUPM-26-8415"
FAVORITE_ARTIST = "Sleeping At Last"
SEED_DIGIT = "5"
ID_SUM = "26"
NAME_LENGTH = "6"
ARTIST_LENGTH = "16"
ARTIST_SUM = "161"

# Generate individualized orders
random.seed(int(SEED_DIGIT) + int(ID_SUM) + int(ARTIST_SUM))
orders = random.sample(range(100, 1000), 10 + int(NAME_LENGTH))
expected = sorted(orders)

datasets = {
    "Sorted": expected.copy(),
    "Reverse": expected[::-1],
    "Random": orders.copy()
}

def bubble(a):
    a = a.copy()
    c = m = 0
    for i in range(len(a) - 1):
        swapped = False
        for j in range(len(a) - 1 - i):
            c += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                m += 1
                swapped = True
        if not swapped:
            break
    return a, c, m

def selection(a):
    a = a.copy()
    c = m = 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            c += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            m += 1
    return a, c, m

def insertion(a):
    a = a.copy()
    c = m = 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            c += 1
            if a[j] > key:
                a[j + 1] = a[j]
                m += 1
                j -= 1
            else:
                break
        a[j + 1] = key
        m += 1
    return a, c, m

algorithms = {
    "Bubble Sort": (bubble, "O(n)", "O(n^2)", "O(n^2)"),
    "Selection Sort": (selection, "O(n^2)", "O(n^2)", "O(n^2)"),
    "Insertion Sort": (insertion, "O(n)", "O(n^2)", "O(n^2)")
}

print(f"Student: {LAST_NAME} | {STUDENT_ID}")
print("Generated Orders:", orders)

for condition, data in datasets.items():
    print(f"\n{condition} Input:", data)
    for name, (algorithm, best, avg, worst) in algorithms.items():
        start = perf_counter()
        result, comparisons, movements = algorithm(data)
        elapsed = perf_counter() - start

        print(f"{name}: Comparisons={comparisons}, Movements={movements}, "
              f"Time={elapsed:.8f}s, Sorted={result == expected}")

print("\nBIG O ANALYSIS")
for name, (_, best, avg, worst) in algorithms.items():
    print(f"{name}: Best={best}, Average={avg}, Worst={worst}")

print("Auxiliary Space: O(1) for all three algorithms")
print("Final Output: Sorting completed; check Sorted=True for all results.")
```
